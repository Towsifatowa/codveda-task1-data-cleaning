## Task 1: Data Cleaning and Preprocessing (Codveda)

Import Libraries

```python
import pandas as pd 
```

Load Dataset

```python
df = pd.read_csv(r"C:\Users\LENOVO\Downloads\3) Sentiment dataset.csv")
# Display first 5 rows
df.head()
```

```python
# see number of rows and columns
print("Shape:", df.shape)
```

```python
# show all the column names
print("\nColumns:")
print(df.columns)

```

```python
# check the data types and non-null values
print("\nDataset Information:")
df.info()
```

```python
# missing Values and duplicates Check
print("Missing Values:")
print(df.isnull().sum())
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# drop unnecessary columns 
df_clean = df.copy()
```

```python
# drop the unnamed column
unnamed_cols = [col for col in df_clean.columns if 'Unnamed' in col]
df_clean = df_clean.drop(columns=unnamed_cols)
print("Columns after drop:", df_clean.columns.tolist())
```

```python
# remove extra space( Standardize text Columns)
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("Standardized Column Names:")
print(df.columns.tolist())
```

```python
# Convert Timestamp to standard datetime format
df_clean['Timestamp'] = pd.to_datetime(df_clean['Timestamp'], errors='coerce')
# extract Year, Month, Day, and Hour from Timestamp
df_clean['Year']  = df_clean['Timestamp'].dt.year
df_clean['Month'] = df_clean['Timestamp'].dt.month
df_clean['Day']   = df_clean['Timestamp'].dt.day
df_clean['Hour']  = df_clean['Timestamp'].dt.hour

print(df_clean[['Timestamp', 'Year', 'Month', 'Day', 'Hour']].head())
```

```python

# Standardize categorical columns

# Standardize Platform values by removing extra spaces and using title case
df_clean['Platform'] = df_clean['Platform'].str.strip().str.title()

# Standardize Sentiment values by removing extra spaces and using title case
df_clean['Sentiment'] = df_clean['Sentiment'].str.strip().str.title()

# Standardize Country values by removing extra spaces and using title case
df_clean['Country'] = df_clean['Country'].str.strip().str.title()

print("\nPlatform value counts:")
print(df_clean['Platform'].value_counts())

print("\nSentiment value counts:")
print(df_clean['Sentiment'].value_counts())

print("\nTop 10 Countries:")
print(df_clean['Country'].value_counts().head(10))
```

```python
# Convert data types
# retweets convert numeric to integer
df_clean['Retweets'] = pd.to_numeric(df_clean['Retweets'], errors='coerce').astype(int)

# likes convert numeric to integer
df_clean['Likes']    = pd.to_numeric(df_clean['Likes'], errors='coerce').astype(int)

# convert year,month,day and hour columns to integer
for col in ['Year', 'Month', 'Day', 'Hour']:
    df_clean[col] = df_clean[col].astype(int)
    
#show the change
print("\nData Types after conversion:")
print(df_clean.dtypes)
```

```python
# Remove duplicate rows after cleaning and standardizing the data
df_clean = df_clean.drop_duplicates()

print("Duplicate rows after removal:", df_clean.duplicated().sum())
```

```python
# Final Data Quality Check
print("Original shape :", df.shape)
print("Cleaned shape  :", df_clean.shape)
print("Missing values :", df_clean.isnull().sum().sum())
print("Duplicates     :", df_clean.duplicated().sum())

print("\nCleaned Data Sample:")
print(df_clean.head())
```

```python
# save the cleaned dataset
df_clean.to_csv("Sentiment_Dataset_Cleaned.csv", index=False)
print("Cleaned dataset saved successfully as 'Sentiment_Dataset_Cleaned.csv'")
```
